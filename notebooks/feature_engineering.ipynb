# Feature Engineering

### Import Libraries

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

### Load Preprocessed Dataset

```python
df = pd.read_csv("../data/preprocessed_news.csv")
df.head()
```

```python
x = df["processed_content"]
y = df["label"]
```

### Apply TF-IDF and convert to Dataframe

```python
vectorizer = TfidfVectorizer(max_features=5000)
```

```python
x_tfidf = vectorizer.fit_transform(x)
```

```python
print(x_tfidf.shape)
```

```python
x_df = pd.DataFrame(x_tfidf.toarray())
x_df.head()
```

### Save Features

```python
x_df.to_csv("../data/x_features.csv", index=False)
y.to_csv("../data/y_labels.csv", index=False)
print("Feature Extraction completed and saved!")
```
